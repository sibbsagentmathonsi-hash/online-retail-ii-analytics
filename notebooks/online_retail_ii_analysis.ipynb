# Online Retail II — guided analytics

This notebook runs the repository's batch-based pipeline against the local workbook and reviews the generated, **aggregated** tables and visualizations. The original workbook is not stored in the repository. See the [README](../README.md) for source, definitions, and setup.

```python
from pathlib import Path
import subprocess
import sys
import pandas as pd
from IPython.display import display, Image

ROOT = Path.cwd()
if ROOT.name == "notebooks":
    ROOT = ROOT.parent
INPUT = ROOT / "data" / "online_retail_II.xlsx"
OUTPUT = ROOT / "outputs"
if not INPUT.is_file():
    raise FileNotFoundError(f"Place the UCI workbook at {INPUT}, or edit INPUT to its local path.")
subprocess.run([sys.executable, str(ROOT / "analytics.py"), "--input", str(INPUT), "--output", str(OUTPUT)], check=True)
print(f"Generated outputs in {OUTPUT}")
```

## Profile, data quality, and headline KPIs

Review duplicate handling and the distinction between positive sales, cancellation invoices, returns, and missing customer IDs before interpreting revenue or customer metrics.

```python
quality = pd.read_csv(OUTPUT / "data_quality.csv")
kpis = pd.read_csv(OUTPUT / "kpi_summary.csv")
display(quality)
display(kpis)
```

## Trend, market, and product analysis

Monthly revenue and country/product rankings are aggregated across both worksheets. Revenue retains the source workbook's price units.

```python
monthly = pd.read_csv(OUTPUT / "monthly_sales.csv")
countries = pd.read_csv(OUTPUT / "country_analysis.csv")
products = pd.read_csv(OUTPUT / "product_analysis.csv")
display(monthly.head(12))
display(countries.head(10))
display(products.head(10))
```

```python
display(Image(OUTPUT / "charts" / "monthly_revenue.png"))
display(Image(OUTPUT / "charts" / "top_countries.png"))
display(Image(OUTPUT / "charts" / "top_products.png"))
```

## Returns, time patterns, RFM segments, and cohorts

Returns are kept separate from positive-sale revenue. RFM and retention use only rows with customer IDs; the exported customer table contains segment aggregates, not customer identifiers.

```python
returns = pd.read_csv(OUTPUT / "returns_analysis.csv")
weekdays = pd.read_csv(OUTPUT / "weekday_analysis.csv")
hours = pd.read_csv(OUTPUT / "hour_analysis.csv")
segments = pd.read_csv(OUTPUT / "customer_segments.csv")
cohorts = pd.read_csv(OUTPUT / "cohort_retention.csv")
display(returns.head(10))
display(weekdays)
display(hours)
display(segments)
display(cohorts.head(20))
display(Image(OUTPUT / "charts" / "weekday_revenue.png"))
display(Image(OUTPUT / "charts" / "cohort_retention.png"))
```
